# Canvas to SVG Converter

This notebook allows you to convert an Obsidian Canvas file (`.canvas`) to an `.svg` file using the `canvas2svg` tool.

Set the input path to your canvas file below.

```python
import os
import subprocess
from pathlib import Path

# 1. Define the input canvas file path here
input_canvas_path = r"C:\Users\Usuario\UGR's Syncthing\Obsidian\LLSS-Obsidian\SWAT nutrient forcing.canvas"

# 2. Path to the canvas2svg script
script_path = os.path.abspath(r"canvas2svg/Code/canvas2svg.py")
```

```python
if not os.path.exists(input_canvas_path):
    print(f"❌ Input file not found: {input_canvas_path}")
else:
    # Determine the output directory and filename
    input_path = Path(input_canvas_path)
    output_svg_path = input_path.with_suffix('.svg')
    
    # Run the script. The script outputs to the current working directory as `{stem}.svg` by default.
    # To ensure it writes exactly where we want, we can change the current working directory
    # to the directory of the input file before running the script.
    command = ["python", script_path, input_path.name]
    
    try:
        result = subprocess.run(command, cwd=input_path.parent, capture_output=True, text=True, check=True)
        print(f"✅ Successfully converted!\nSaved SVG to: {output_svg_path}")
        if result.stdout:
            print("\nOutput:", result.stdout)
    except subprocess.CalledProcessError as e:
        print("❌ Error during conversion.")
        print(e.stderr)
        print(e.stdout)
```
